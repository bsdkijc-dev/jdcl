# 01 — Exploratory Analysis (Synthetic JFLM 2027)

Explores synthetic correlations across `region_tier`, `service_channel`, and `vulnerability_group`. Correlations in synthetic data reflect modeling assumptions, not causal empirical claims.


```python
import pandas as pd
df = pd.read_csv('../data/synthetic/jflm_2027.csv')
print('By Region Tier:')
print(df.groupby('region_tier')[['distance_to_court_km', 'resolution_days', 'service_completed']].mean().round(2))
print('\nMissingness counts (zero-imputation preserved):')
print(df.isna().sum()[df.isna().sum() > 0])
```
