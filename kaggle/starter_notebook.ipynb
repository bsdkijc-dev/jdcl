# JFLM 2027 — Kaggle Community Competition Starter Notebook

Generates a valid baseline submission (`predicted_resolution_days`) on the synthetic holdout split.


```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
from jdcl.models.baselines import deterministic_split, TabularPreprocessor, RidgeLinearBaseline
from jdcl.competition.engine import CompetitionEngine
df = pd.read_csv('../data/synthetic/jflm_2027.csv')
split = deterministic_split(df, seed=20270101)
prep = TabularPreprocessor().fit(split.train_df)
model = RidgeLinearBaseline().fit(prep.transform(split.train_df), split.train_df['resolution_days'].to_numpy(float))
sub = pd.DataFrame({'record_id': split.test_df['record_id'], 'predicted_resolution_days': model.predict(prep.transform(split.test_df))})
res = CompetitionEngine('jflm-2027').evaluate_submission(split.test_df, sub, sub)
print('Valid:', res['valid'], '| Reproducible:', res['reproducible'], '| Metrics:', res['metrics'])
```
